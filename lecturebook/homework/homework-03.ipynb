## Problem 1 — Maximum entropy and Bayesian sensor calibration (25 points)

A sensor's **calibration residual** is its reported value minus a known
reference value. Let $b$, measured in millivolts (mV), denote the sensor's
unknown but fixed signed zero offset. Before collecting calibration
measurements, the available engineering information is

$$
b\in\mathbb{R},
\qquad
\mathbb{E}[b]=b_0,
\qquad
\mathbb{V}[b]=\tau_0^2.
$$

Assume $\tau_0>0$.

Support and moments do not uniquely determine a prior distribution. Once
the variable and reference density are specified, the maximum-entropy
principle selects the feasible distribution that introduces the least
additional structure beyond the stated information.

### 1.1 Derive the maximum-entropy prior (7 points)

Take the reference density to be $q(b)=1$ relative to Lebesgue measure—that
is, ordinary integration with respect to $db$—in the stated mV coordinate.
Equivalently, write the variance constraint as the second-moment constraint

$$
\mathbb{E}[b^2]=b_0^2+\tau_0^2.
$$

Lecture 11's maximum-entropy result gives a density of the form

$$
\pi(b)
=
\frac{1}{Z}
\exp\left(\eta_1b+\eta_2b^2\right),
\qquad b\in\mathbb{R},

where $Z$ is the normalization constant.

3. Determine $Z$ and identify the resulting named prior distribution.
   **(2 points)**

You may use

### Solution: Derive the maximum entropy prior

1. Since $b \in \mathbb{R}$, the density must be integrable over the entire real line. As $|b| \to \infty$, the quadratic term $\eta_2 b^2$ dominates the exponent. Therefore, $\eta_2 < 0$ is required so that the exponent approaches $-\infty$, ensuring that $\pi(b)$ decays to zero and the normalization constant $Z$ remains finite.> **Response:** Replace this text with your derivation.

2. Starting with

$$
\pi(b) \propto \exp(\eta_1 b + \eta_2 b^2),
$$

complete the square in the exponent:

$$
\eta_1 b + \eta_2 b^2
=
\eta_2\left(b^2+\frac{\eta_1}{\eta_2}b\right)
=
\eta_2\left(b+\frac{\eta_1}{2\eta_2}\right)^2
-\frac{\eta_1^2}{4\eta_2}.
$$

Comparing with the Gaussian form

$$
\exp\left[-\frac{(b-\mu)^2}{2\sigma^2}\right],
$$

we identify

$$
\mu=-\frac{\eta_1}{2\eta_2},
\qquad
\sigma^2=-\frac{1}{2\eta_2}.
$$

Using the first moment constraint,

$$
\mathbb{E}[b]=b_0,
$$

gives

$$
-\frac{\eta_1}{2\eta_2}=b_0.
$$

Using the second moment constraint,

$$
\mathbb{E}[b^2]=b_0^2+\tau_0^2,
$$

and the fact that

$$
\mathbb{E}[b^2]=\mu^2+\sigma^2,
$$

we obtain

$$
b_0^2-\frac{1}{2\eta_2}
=
b_0^2+\tau_0^2.
$$

Therefore,

$$
\eta_2=-\frac{1}{2\tau_0^2}.
$$

Substituting this into the first moment constraint gives

$$
\eta_1=\frac{b_0}{\tau_0^2}.
$$

Thus,

$$
\eta_1=\frac{b_0}{\tau_0^2},
\qquad
\eta_2=-\frac{1}{2\tau_0^2}.
$$

3. Starting with

$$
\pi(b)=\frac{1}{Z}\exp\left(
\frac{b_0}{\tau_0^2}b-\frac{b^2}{2\tau_0^2}
\right),
$$

completing the square gives

$$
\pi(b)
=
\frac{1}{Z}
\exp\left(\frac{b_0^2}{2\tau_0^2}\right)

Comparing with the normalized Gaussian density,

$$
\frac{1}{\sqrt{2\pi\tau_0^2}}
\exp\left[-\frac{(b-b_0)^2}{2\tau_0^2}\right],
$$

the normalization factors must satisfy

$$
\frac{1}{Z}\exp\left(\frac{b_0^2}{2\tau_0^2}\right)
=
\frac{1}{\sqrt{2\pi\tau_0^2}}.
$$

Therefore,

$$
Z=\sqrt{2\pi\tau_0^2}
\exp\left(\frac{b_0^2}{2\tau_0^2}\right).
$$

The normalized prior is therefore

$$
\pi(b)=
\frac{1}{\sqrt{2\pi\tau_0^2}}
\exp\left[-\frac{(b-b_0)^2}{2\tau_0^2}\right],
$$

which is a Gaussian distribution with mean $b_0$ and variance $\tau_0^2$.


### 1.2 Derive the analytical posterior (10 points)

Suppose the calibration residuals $Y_1,\ldots,Y_N$ are conditionally
independent given $b$ and satisfy

$$
Y_i\mid b\sim N(b,\sigma^2),
\qquad i=1,\ldots,N,
$$

where $\sigma>0$, $N\geq1$, and the measurement-noise variance $\sigma^2$ is
known. Let $y_{1:N}=(y_1,\ldots,y_N)$ denote the observed residuals, and
define

$$
\bar y=\frac{1}{N}\sum_{i=1}^N y_i.
$$

1. Write the likelihood as a function of $b$, retaining all factors that
   depend on $b$. **(2 points)**
2. Starting from Bayes' rule,

   $$
   p(b\mid y_{1:N})

   $$

   average of $b_0$ and $\bar y$ whose weights are determined by prior
   precision and data precision. Verify that the weights sum to one and
   interpret them. **(2 points)**
4. A prior family is **conjugate** to a likelihood when the posterior belongs
   to the same distribution family as the prior. Explain why this update is

You may use

=

Leave the posterior in terms of $N$ and $\bar y$; no numerical observations

### Solution.
1.
Since the observations are conditionally independent given $b$, the likelihood is the product of the individual Gaussian densities:

$$
p(y_{1:N}\mid b)
=
\prod_{i=1}^{N} p(y_i\mid b).
$$

Since

$$
Y_i\mid b \sim \mathcal{N}(b,\sigma^2),
$$

the likelihood becomes

$$
p(y_{1:N}\mid b)
=
\prod_{i=1}^{N}
\frac{1}{\sqrt{2\pi\sigma^2}}
\exp\left[-\frac{(y_i-b)^2}{2\sigma^2}\right].
$$

Combining the product,

$$
p(y_{1:N}\mid b)
=
(2\pi\sigma^2)^{-N/2}
\exp\left[
-\frac{1}{2\sigma^2}
\sum_{i=1}^{N}(y_i-b)^2
\right].
$$

Since the prefactor does not depend on $b$, the likelihood as a function of $b$ can be written as

$$
p(y_{1:N}\mid b)
\propto
\exp\left[
-\frac{1}{2\sigma^2}
\sum_{i=1}^{N}(y_i-b)^2
\right].
$$

2. Starting from Bayes' rule,

$$
p(b\mid y_{1:N}) \propto p(y_{1:N}\mid b)\pi(b).
$$

Using

$$
\sum_{i=1}^{N}(y_i-b)^2
=
\sum_{i=1}^{N}(y_i-\bar y)^2+N(b-\bar y)^2,
$$

the first term does not depend on $b$, so the likelihood can be written as

$$
p(y_{1:N}\mid b)
\propto
\exp\left[-\frac{N}{2\sigma^2}(b-\bar y)^2\right].
$$

From Part 1.1, the prior is

$$
\pi(b)
\propto
\exp\left[-\frac{(b-b_0)^2}{2\tau_0^2}\right].
$$

Multiplying the likelihood and prior gives

$$
p(b\mid y_{1:N})
\propto
\exp\left[
-\frac{N}{2\sigma^2}(b-\bar y)^2
-\frac{1}{2\tau_0^2}(b-b_0)^2
\right].
$$

Expanding the terms involving $b$,

$$
p(b\mid y_{1:N})
\propto
\exp\left[
-\frac{1}{2}
\left(
\left(\frac{N}{\sigma^2}+\frac{1}{\tau_0^2}\right)b^2
-
2\left(\frac{N\bar y}{\sigma^2}+\frac{b_0}{\tau_0^2}\right)b
\right)
\right].
$$

Define

$$
A=\frac{N}{\sigma^2}+\frac{1}{\tau_0^2},
\qquad
B=\frac{N\bar y}{\sigma^2}+\frac{b_0}{\tau_0^2}.
$$

Then

$$
Ab^2-2Bb
=
A\left(b-\frac{B}{A}\right)^2-\frac{B^2}{A}.
$$

The final term is independent of $b$ and can be absorbed into the normalization constant. Therefore,

$$
p(b\mid y_{1:N})
\propto
\exp\left[
-\frac{A}{2}
\left(b-\frac{B}{A}\right)^2
\right].
$$

Comparing with a Gaussian density,

$$
b\mid y_{1:N}\sim\mathcal{N}(b_N,v_N),
$$

where

$$
v_N
=
\left(
\frac{1}{\tau_0^2}+\frac{N}{\sigma^2}
\right)^{-1},
$$

and

$$
b_N
=
v_N
\left(
\frac{b_0}{\tau_0^2}
+
\frac{N\bar y}{\sigma^2}
\right).
$$

3. The posterior mean is

$$
b_N
=
v_N
\left(
\frac{b_0}{\tau_0^2}
+
\frac{N\bar y}{\sigma^2}
\right),
$$

where

$$
v_N
=
\left(
\frac{1}{\tau_0^2}+\frac{N}{\sigma^2}
\right)^{-1}.
$$

Substituting $v_N$ gives

$$
b_N
=
\frac{\frac{1}{\tau_0^2}}
{\frac{1}{\tau_0^2}+\frac{N}{\sigma^2}}b_0
+
\frac{\frac{N}{\sigma^2}}
{\frac{1}{\tau_0^2}+\frac{N}{\sigma^2}}\bar y.
$$

Therefore, defining the weights

$$
w_0=
\frac{\frac{1}{\tau_0^2}}
{\frac{1}{\tau_0^2}+\frac{N}{\sigma^2}},
\qquad
w_y=
\frac{\frac{N}{\sigma^2}}
{\frac{1}{\tau_0^2}+\frac{N}{\sigma^2}},
$$

we have

$$
b_N=w_0b_0+w_y\bar y.
$$

The weights sum to one since

$$
w_0+w_y
=
\frac{\frac{1}{\tau_0^2}+\frac{N}{\sigma^2}}
{\frac{1}{\tau_0^2}+\frac{N}{\sigma^2}}
=1.
$$

Thus, the posterior mean is a weighted average of the prior mean $b_0$ and the observed sample mean $\bar y$. A larger prior precision $1/\tau_0^2$ places more weight on $b_0$, while a larger data precision $N/\sigma^2$ places more weight on $\bar y$.

4. The prior distribution is Gaussian,

$$
b\sim\mathcal{N}(b_0,\tau_0^2),
$$

and the posterior distribution obtained after observing the Gaussian measurements is also Gaussian,

$$
b\mid y_{1:N}\sim\mathcal{N}(b_N,v_N).
$$

Therefore, the Gaussian prior is conjugate to the Gaussian likelihood (with known variance), since the posterior belongs to the same distribution family as the prior.

### Solution:

$$

\left(

$$

We require

$$
v_N \leq v_*,
$$

so

$$
\left(
\frac{1}{\tau_0^2}+\frac{N}{\sigma^2}
\right)^{-1}

Since both sides are positive, taking reciprocals reverses the inequality:

$$
\frac{1}{\tau_0^2}+\frac{N}{\sigma^2}
\geq \frac{1}{v_*}.

$$
N
\geq

\left(
\frac{1}{v_*}-\frac{1}{\tau_0^2}
\right).

Since $N$ must be a nonnegative integer,

$$
N_{\min}
=
\max\left[

\left\lceil

\left(

### 1.4 Predict a future calibration residual (3 points)

Let $Y_*$ denote a future residual from the same sensor, with

$$
Y_*\mid b\sim N(b,\sigma^2),
$$

independently of the calibration data given $b$. The **posterior predictive
distribution** describes an unobserved future measurement after averaging
over posterior uncertainty in $b$.

1. Derive the posterior predictive distribution of $Y_*$ and its central 95%
   predictive interval. You may use
   $P(-1.96\leq Z\leq1.96)\approx0.95$ for $Z\sim N(0,1)$.
   For the numerical design at $N=15$, also report the predictive variance
   and the 95% predictive half-width. Because no numerical value of
   $\bar y$ is supplied, leave this numerical interval centered at the
   symbolic posterior mean $b_{15}$.
   **(2 points)**
2. Interpret the two contributions to its variance. State which contribution
   can be reduced by collecting more calibration measurements and identify
   the limiting predictive variance as $N\to\infty$. **(1 point)**

**Problem 1 credit.** The maximum-entropy construction follows
[Jaynes (1957)](https://doi.org/10.1103/PhysRev.106.620) and Lecture 11. The
Normal–Normal update is standard conjugate Bayesian analysis. The sensor
calibration setting, sample-size design question, and numerical values were
developed for this course.


### 1. Posterior predictive distribution

From Part 1.2,

$$
b\mid y_{1:N}\sim\mathcal{N}(b_N,v_N),
$$

where

$$
b_N
=
v_N\left(
\frac{b_0}{\tau_0^2}
+
\frac{N\bar y}{\sigma^2}
\right),
\qquad
v_N
=
\left(
\frac{1}{\tau_0^2}+\frac{N}{\sigma^2}
\right)^{-1}.
$$

A future calibration residual satisfies

$$
Y_*\mid b\sim\mathcal{N}(b,\sigma^2).
$$

So, we can write

$$
Y_*=b+\epsilon_*,
\qquad
\epsilon_*\sim\mathcal{N}(0,\sigma^2).
$$

Averaging over the posterior uncertainty in $b$, the posterior predictive distribution is

$$
Y_*\mid y_{1:N}
\sim
\mathcal{N}\left(b_N,\sigma^2+v_N\right).
$$

Therefore, the central $95\%$ posterior predictive interval is

$$
b_N
\pm
1.96\sqrt{\sigma^2+v_N}.
$$

For the numerical design from Part 1.3,

$$
N=15,
\qquad
\tau_0=1.00\text{ mV},
\qquad
\sigma=1.00\text{ mV}.
$$

The posterior variance is

$$
v_{15}
=
\left(
\frac{1}{1^2}+\frac{15}{1^2}
\right)^{-1}
=
\frac{1}{16}
=
0.0625\text{ mV}^2.
$$

Hence, the predictive variance is

$$
\sigma^2+v_{15}
=
1+0.0625
=
1.0625\text{ mV}^2.
$$

The $95\%$ predictive half-width is

$$
1.96\sqrt{1.0625}
\approx
2.02\text{ mV}.
$$

Since no numerical value of $\bar y$ is given, the interval remains centered at the symbolic posterior mean $b_{15}$:

$$
b_{15}\pm2.02\text{ mV}.
$$

### 2. Interpretation of the predictive variance

The predictive variance is

$$
\operatorname{Var}(Y_*\mid y_{1:N})
=
\sigma^2+v_N.
$$

The term $\sigma^2$ represents the measurement noise in a future residual, while $v_N$ represents the posterior uncertainty in the unknown sensor offset $b$.

Collecting more calibration measurements reduces $v_N$, since

$$
v_N=
\left(
\frac{1}{\tau_0^2}+\frac{N}{\sigma^2}
\right)^{-1}
\rightarrow 0
\qquad\text{as }N\rightarrow\infty.
$$

However, the future measurement noise $\sigma^2$ cannot be reduced by collecting more calibration measurements. Therefore,

$$
\lim_{N\rightarrow\infty}
\operatorname{Var}(Y_*\mid y_{1:N})
=
\sigma^2.
$$

Thus, the limiting predictive variance is the measurement-noise variance $\sigma^2$.

## Problem 2 — Bayesian earthquake-rate study (75 points)

The course's frozen [USGS ComCat](https://earthquake.usgs.gov/data/comcat/)
snapshot contains catalog events with magnitude at least 6.5 from January
1, 1900 through December 31, 2025 in the rectangular study region
$32\le\text{latitude}\le37$ degrees N and
$-122\le\text{longitude}\le-114$ degrees. This rectangle includes parts of
California and Baja California; it is not a Southern-California boundary.

Let $\Delta_t=1$ year denote the known exposure for calendar year $t$.
Model the annual counts as conditionally independent observations

$$
Y_t\mid\lambda,\Delta_t
\sim\operatorname{Poisson}(\lambda\Delta_t),
$$

where $\lambda$ is a constant event rate in events/year. This is a compact
teaching model, not an official seismic-hazard forecast.

**Parameterization note.** Throughout the notebook, Gamma distributions use
shape $\alpha$ and rate $\beta$. SciPy instead expects
`stats.gamma(a=alpha, scale=1 / beta)`.

For $\lambda>0$, the shape-rate density and its moments are

$$
f(\lambda\mid\alpha,\beta)
=
\frac{\beta^\alpha}{\Gamma(\alpha)}
\lambda^{\alpha-1}e^{-\beta\lambda},
\qquad
\mathbb{E}[\lambda]=\frac{\alpha}{\beta},
\qquad
\mathbb{V}[\lambda]=\frac{\alpha}{\beta^2}.
$$

**Gamma–Poisson update card.** If
$\lambda\sim\operatorname{Gamma}(\alpha_0,\beta_0)$, total observed count is
$n=\sum_t y_t$, and total known exposure is $E=\sum_t\Delta_t$, then

$$
\lambda\mid\mathbf y,\boldsymbol\Delta
\sim\operatorname{Gamma}(\alpha_N,\beta_N),
